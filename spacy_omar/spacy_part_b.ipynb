```python
import spacy
from spacy.tokens import Span
```

```python
 #!python -m spacy download en_core_web_md
```

```python
nlp = spacy.load("en_core_web_md")

with open('mobydick.txt', 'r', encoding='utf-8', errors='ignore') as file:
    moby = file.read()

with open('ai_forecast1.txt', 'r', encoding='utf-8', errors='ignore') as file:
    ai_forcast1 = file.read()

with open('ai_forecast2.txt', 'r', encoding='utf-8', errors='ignore') as file:
    ai_forcast2 = file.read()

moby_doc = nlp(moby)
ai_forecast1_doc = nlp(ai_forcast1)
ai_forecast2_doc = nlp(ai_forcast2)
```

```python
print("Similarity between Moby Dick and AI Forecast 1:", moby_doc.similarity(ai_forecast1_doc))
print("Similarity between Moby Dick and AI Forecast 2:", moby_doc.similarity(ai_forecast2_doc))
print("Similarity between AI Forecast 1 and AI Forecast 2:", ai_forecast1_doc.similarity(ai_forecast2_doc))
```

```python
print("First 100 tokens comparison:")
print("Similarity between Moby Dick and AI Forecast 1:", moby_doc[:100].similarity(ai_forecast1_doc[:100]))
print("Similarity between Moby Dick and AI Forecast 2:", moby_doc[:100].similarity(ai_forecast2_doc[:100]))
print("Similarity between AI Forecast 1 and AI Forecast 2:", ai_forecast1_doc[:100].similarity(ai_forecast2_doc[:100]))
```

```python
nlp_blank = spacy.blank("en")

text = "Omar Hammad is a professional DevOps."
doc_blank = nlp_blank(text)

span = Span(doc_blank, 0, 2, "PERSON")
doc_blank.ents = list(doc_blank.ents) + [span]
for ent in doc_blank.ents:
    print(f"Entity: {ent.text}, Label: {ent.label_}")
```
