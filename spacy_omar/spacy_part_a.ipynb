```python
import spacy
from spacy.matcher import Matcher
```

```python
nlp = spacy.load("en_core_web_sm")

with open('mobydick.txt', 'r', encoding='utf-8', errors='ignore') as file:
    moby = file.read()

with open('ai_forecast1.txt', 'r', encoding='utf-8', errors='ignore') as file:
    ai_forcast1 = file.read()

with open('ai_forecast2.txt', 'r', encoding='utf-8', errors='ignore') as file:
    ai_forcast2 = file.read()

moby_doc = nlp(moby)
ai_forcast1_doc = nlp(ai_forcast1)
ai_forcast2_doc = nlp(ai_forcast2)
```

```python
def analyze_text(doc):
    for token in doc:
        print(f"Token: {token.text:<15} Pos: {token.pos_}({spacy.explain(token.pos_) + ')':<50} Dep {token.dep_}")

    print("=" * 100)
    for sent in doc.sents:
        print(f"{sent}\n")

    print("=" * 100)
    for ent in doc.ents:
        print(f"Entity: {ent.text:<40} Label: {ent.label_}")

    print("=" * 100)
    pattern = [{'TEXT': "Artificial"}, {'TEXT': 'Intelligence'}]
    matcher = Matcher(nlp.vocab)
    matcher.add("ARTIFICIAL_INTELLIGENCE", [pattern])
    matches = matcher(doc)
    print(f"Total Matches for Artificial Intelligence ({len(matches)})")
    for match_id, start, end in matches:
        span = doc[start:end]
        print(f"Match found:{span.sent}")

    print("=" * 100)
    pattern = [{'TEXT': "AI"}, {'POS': 'VERB'}]
    matcher = Matcher(nlp.vocab)
    matcher.add("ARTIFICIAL_INTELLIGENCE_FOLLOWED_BY_VERB", [pattern])
    matches = matcher(doc)
    print(f"Total Matches for Artificial Intelligence + VERB ({len(matches)})")
    for match_id, start, end in matches:
        span = doc[start:end]
        print(f"Match found:  {span.text}")


    print("=" * 100)
    pattern = [{'LIKE_NUM':True}, {'TEXT': '%'}]
    matcher = Matcher(nlp.vocab)
    matcher.add("PERCENT", [pattern])
    matches = matcher(doc)
    print(f"Total Matches for Percentage ({len(matches)})")
    for match_id, start, end in matches:
        span = doc[start:end]
        print(f"Match found:  {span.text}")

    print("=" * 100)
    pattern = [{'ENT_TYPE': "ORG"}]
    matcher = Matcher(nlp.vocab)
    matcher.add("COMPANIES", [pattern])
    matches = matcher(doc)
    print(f"Total Matches for Company Names ({len(matches)})")
    for match_id, start, end in matches:
        span = doc[start:end]
        print(f"Match found:  {span.text}")

```

```python
analyze_text(moby_doc)
```

```python
analyze_text(ai_forcast1_doc)
```

```python
analyze_text(ai_forcast2_doc)
```
