```python
import spacy
from spacy import Language
```

```python
nlp = spacy.load("en_core_web_sm")
print(nlp.pipe_names)
```

```python
@Language.component("longest_token_comp")
def longest_doc(doc):
    longest_token = max(doc, key=len)
    print(f" - This token '{longest_token.text}' is the longest length of ({len(longest_token)})")
    return doc

nlp.add_pipe('longest_token_comp',after='tok2vec')
print(nlp.pipe_names)
```
