```python
df['quality'].value_counts().sort_values().plot(kind='bar', color='skyblue', figsize=(8, 5))

plt.xlabel('Quality', fontsize=10)
plt.ylabel('Count', fontsize=12)
plt.title('Count of Each Wine Quality Rating (Smallest to Largest)', fontsize=14)

plt.show()
```

```python
"""
fixed acidity: sharp crisps taste

 Refers to non-volatile acids present in the wine, which do not evaporate easily during fermentation.
 Common fixed acids include tartaric, malic, and citric acids.
 Unit: grams per liter (g/L)
 Importance: Contributes to the wine's ""tartness"" and ""structure"".
 Higher fixed acidity gives the wine a crisp and refreshing taste.
"""

grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "fixed acidity", bins=8, color='red')
plt.show()
```

```python
"""
volatile acidity: vinegar-like taste low quality

Refers to the amount of volatile acids in the wine, primarily acetic acid. These acids can evaporate easily, 
and high levels of volatile acidity can give the wine an unpleasant vinegar-like taste.

Unit: grams per liter (g/L)
Importance: High volatile acidity negatively affects wine quality, as it can lead to off-flavors.
"""

grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "volatile acidity", bins=8, color='red')
plt.show()
```

```python
"""
citric acid: freshness and brightness in flavor

Citric acid is a minor acid in wine, adding a small amount of freshness and brightness to the flavor.
Unit: grams per liter (g/L)
Importance: It helps balance sweetness and acidity in the wine.
"""
grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "citric acid", bins=8, color='red')
plt.show()
```

```python
"""
residual sugar: sweetness of the wine

Residual sugar refers to the amount of sugar left in the wine after fermentation is complete.
Unit: grams per liter (g/L)
Importance: It indicates the sweetness of the wine. Lower levels indicate a dry wine, while higher levels indicate a sweeter wine.
"""

grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "residual sugar", bins=8, color='red')
plt.show()
```

```python
"""
chlorides: salt content

Represents the salt content in the wine, often from chloride ions.
Unit: grams per liter (g/L)
Importance: High chloride content can give the wine a salty taste, which is generally undesirable.
"""
grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "chlorides", bins=8, color='red')
plt.show()
```

```python
"""
free sulfur dioxide: antimicrobial and antibacterial  (potassium metabisulfite)

The portion of sulfur dioxide (SO₂) that remains unbound and is free to act as an antimicrobial and antioxidant.
Unit: milligrams per liter (mg/L)
Importance: Helps preserve the wine and prevent oxidation.
"""
grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "free sulfur dioxide", bins=8, color='red')
plt.show()
```

```python
"""
total sulfur dioxide: preservative

The total amount of both free and bound sulfur dioxide in the wine.
Unit: milligrams per liter (mg/L)
Importance: Sulfur dioxide is used to prevent spoilage, but excessive amounts can affect flavor.
"""
grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "total sulfur dioxide", bins=8, color='red')
plt.show()
```

```python
"""
density: weight of the wine, sugar and alcohol content, viscosity

The wine's density, which is influenced by its sugar and alcohol content. It is closely related to the specific gravity of the wine.
Unit: grams per cubic centimeter (g/cm³)
Importance: Helps determine alcohol content and sugar levels during fermentation.
"""
grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "density", bins=8, color='red')
plt.show()
```

```python
"""
pH: acidity of the wine

Measures the acidity of the wine, with lower values indicating higher acidity.
Unit: pH scale (0–14)
Importance: Affects the wine's taste and stability. The pH level is crucial for overall flavor balance and preservation.
"""
grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "pH", bins=8, color='red')
plt.show()
```

```python
"""
sulphates: preservative, mouthfeel

Refers to the concentration of sulfate compounds (like potassium sulfate) in the wine.
Unit: grams per liter (g/L)
Importance: Sulphates contribute to wine preservation and stability, and they can affect mouthfeel.
"""

grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "sulphates", bins=8, color='red')
plt.show()
```

```python
"""
alcohol: body and sensory profile

The alcohol content of the wine.
Unit: percentage (%)
Importance: Alcohol contributes to the wine’s body and overall sensory profile. Higher alcohol levels are often found in fuller-bodied wines.
"""

grid = sns.FacetGrid(df, col="quality", height=3, aspect=1) 
grid.map(plt.hist, "alcohol", bins=8, color='red')
plt.show()
```
