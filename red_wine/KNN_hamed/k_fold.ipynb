```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import KFold
```

```python
df = pd.read_csv('../winequality-red-clean.csv')
```

```python
df.sample(5)
```

### K-Fold Cross Validation with KNN

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop('quality', axis=1), df['quality'], test_size=0.33, random_state=42)
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
knn.score(X_test, y_test)
```

```python
kf = KFold(n_splits=5)
kf
```

```python
for train_index, test_index in kf.split(df):
    print("TRAIN:", train_index, "TEST:", test_index)
    X_train, X_test = df.drop('quality', axis=1).iloc[train_index], df.drop('quality', axis=1).iloc[test_index]
    y_train, y_test = df['quality'].iloc[train_index], df['quality'].iloc[test_index]
    knn = KNeighborsClassifier(n_neighbors=9)
    knn.fit(X_train, y_train)
    print(knn.score(X_test, y_test))
```

## Rapid and straightforward way to do the same thing

```python
from sklearn.model_selection import cross_val_score

knn = KNeighborsClassifier(n_neighbors=9)
scores = cross_val_score(knn, df.drop('quality', axis=1), df['quality'], cv=5)
scores
```
