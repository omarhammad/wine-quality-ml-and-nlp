### Moby Dick text exploration with named entities and part of speach POS

###### tokenizing a document

```python
import spacy
from spacy.matcher import Matcher
nlp = spacy.load("en_core_web_sm")
matcher = Matcher(nlp.vocab)

# optionally we can set the max length of the text that we are going to process in this case the moby dick is below 1.5 million characters
nlp.max_length = 1500000  

with open('mobydick.txt', 'r', encoding='utf-8') as file:
    moby_dick_text = file.read()

# the text can also be broken down into many chunks and then fed into the model
chunk_size = 100000

for i in range(0, len(moby_dick_text), chunk_size):
    chunk = moby_dick_text[i:i + chunk_size]
    doc = nlp(chunk)
```

###### checking the attributes related to each token object

```python
"""
Tokens have different attributes and they are described below:
- POS: Part of speech represents the role the word plays in the sentence, e.g., verb, noun, proper noun, etc.
- Lemma: Represents the base or stem of the word, e.g., 'called' -> 'call'.
- Dependency: The grammatical relationship between this word and other words in the sentence. It explains the syntactic role the word plays (subject, object, modifier, etc.).
- Head: The headword in the sentence to which the current token is related. It shows the word that this token syntactically depends on.
"""

for token in doc:
    print(f"Token: {token.text}, POS: {token.pos_}, Lemma: {token.lemma_}, Dependency: {token.dep_}, Head: {token.head.text}")
```

###### looping over the named entities and printing them

```python
for ent in doc.ents:
    print(ent.text)
```
