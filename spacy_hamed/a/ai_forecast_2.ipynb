### importing spacy and the text file and tokenizing the text file

```python
import spacy

#  at this step we need to download a model and for this exercise we are going to download the medium sized model
nlp = spacy.load("en_core_web_sm")

# optionally we can set the max length of the text that we are going to process in this case the moby dick is below 1.5 million characters
nlp.max_length = 5000000  

with open('ai_forecast2.txt', 'r', encoding='utf-8', errors='replace') as file:
    ai_forest = file.read()


# the text can also be broken down into many chunks and then fed into the model
chunk_size = 100000

for i in range(0, len(ai_forest), chunk_size):
    chunk = ai_forest[i:i + chunk_size]
    doc = nlp(chunk)


# at this stage we have a model that has consumed our text already and tokenized it, next step we can see the tokens and their attributes if needed
```

##### im looking for a pattern that finds word AI followed by a verb.

```python
from spacy.matcher import Matcher
matcher = Matcher(nlp.vocab)
pattern = [{"TEXT": "AI"}, {"POS": "VERB"}]

matcher.add("AI_VERB_PATTERN", [pattern])

matches = matcher(doc)

print("Matches for 'AI' followed by a verb:")
for match_id, start, end in matches:
    matched_span = doc[start:end]
    print(f"Found match: {matched_span.text}, at position: {start}-{end}")
```

##### im looking for a pattern that finds a number followed by a % sign

```python
from spacy.matcher import Matcher
matcher = Matcher(nlp.vocab)

pattern = [{"LIKE_NUM": True}, {"TEXT": "%"}]

matcher.add("NUMBER_PERCENT_PATTERN", [pattern])

matches = matcher(doc)

print("Matches for numbers followed by %:")
for match_id, start, end in matches:
    matched_span = doc[start:end]  # Create a span from the start and end positions
    print(f"Found match: {matched_span.text}, at position: {start}-{end}")
```

##### im looking for a pattern that find a organization name

```python
pattern = [{"ENT_TYPE": "ORG"}]
matcher.add("COMPANY_NAME_PATTERN", [pattern])

matches = matcher(doc)

print("Matches for company names:")
for match_id, start, end in matches:
    matched_span = doc[start:end]  # Create a span from the start and end positions
    print(f"Found match: {matched_span.text}, at position: {start}-{end}")
```
