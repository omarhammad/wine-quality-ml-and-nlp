## compairing the similariries between the data ai_forecast1 and the data ai_forecast_2 and moby_dick

##### lets firs import the spacy model and then import the text files 
##### nlp max length is the maximum number of charactors for tokenization
##### chunk size is the number of charactors that each chunk has, a text can be broken down to many equal chunks

```python
import spacy
from spacy.matcher import Matcher
nlp = spacy.load("en_core_web_md")
matcher = Matcher(nlp.vocab)
nlp.max_length = 1500000 
chunk_size = 10000
```

##### breaking moby dick text into many chunks using a for loop

```python
with open('../a/mobydick.txt', 'r', encoding='utf-8', errors='ignore') as file:
    moby_dick_text = file.read()
    
for i in range(0, len(moby_dick_text), chunk_size):
    chunk = moby_dick_text[i:i + chunk_size]
    doc_moby = nlp(chunk)    
```

##### breaking ai forecast 1 text file into many chunks using a for loop

```python
with open('../a/ai_forecast1.txt', 'r', encoding='utf-8', errors='ignore') as file:
    ai_1 = file.read()
    
for i in range(0, len(ai_1), chunk_size):
    chunk = ai_1[i:i + chunk_size]
    doc_ai_1 = nlp(chunk)  
```

##### breaking ai forecast 1 text file into many chunks using a for loop

```python
with open('../a/ai_forecast2.txt', 'r', encoding='utf-8', errors='ignore') as file:
    ai_2 = file.read()

for i in range(0, len(ai_2), chunk_size):
    chunk = ai_2[i:i + chunk_size]
    doc_ai_2 = nlp(chunk)  
```

##### checking the similaryty of the moby dick and ai 1

```python
sim_moby_ai1 = doc_moby.similarity(doc_ai_1)
sim_moby_ai1
```

##### checking the similaryty of the moby dick and ai 2

```python
sim_moby_ai2 = doc_moby.similarity(doc_ai_2)
sim_moby_ai2
```

##### checking the similaryty of the ai 1 and ai 2

```python
sim_ai2_ai1 = doc_ai_2.similarity(doc_ai_1)
sim_ai2_ai1
```

##### in the next cell I want to add my name to the spacy pipeline as a named entity

```python
import spacy
from spacy.tokens import Span

nlp = spacy.blank("en")

ner = nlp.add_pipe('ner', last=True)
ner.add_label('PERSON')

nlp.initialize()

test_text = "Hamed Mahdavi Far is a software developer."
doc = nlp(test_text)

span = Span(doc, 0, 3, label="PERSON")
doc.ents = [span]

for ent in doc.ents:
    print(ent.text, ent.label_)
```
