```python
import spacy
from spacy.language import Language

nlp = spacy.load("en_core_web_sm")

print("Pipeline components:", nlp.pipe_names)

@Language.component("longest_token_finder")
def custom_component(doc):
    longest_token = max(doc, key=len)
    print(f"Longest token: {longest_token}")
    return doc

nlp.add_pipe("longest_token_finder", after="ner")

doc = nlp("Hamed Mahdavi Far is a software developer working on various projects.")

print("Pipeline components:", nlp.pipe_names)

```
