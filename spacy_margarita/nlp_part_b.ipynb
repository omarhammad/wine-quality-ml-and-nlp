```python
import spacy
nlp = spacy.load("en_core_web_md")
```

```python
with open('text1.txt', 'r', encoding='utf-8', errors='ignore') as file:
    text1 = file.read()
    
with open('text2.txt', 'r', encoding='utf-8', errors='ignore') as file:
    text2 = file.read()

with open('text3.txt', 'r', encoding='utf-8', errors='ignore') as file:
    text3 = file.read()
    
    
doc1 = nlp(text1)
doc2 = nlp(text2)
doc3 = nlp(text3)
```

```python
# Compare doc1 with doc2
similarity_doc1_doc2 = doc1.similarity(doc2)
print(f"Similarity between doc1 and doc2: {similarity_doc1_doc2}")

# Compare doc1 with doc3
similarity_doc1_doc3 = doc1.similarity(doc3)
print(f"Similarity between doc1 and doc3: {similarity_doc1_doc3}")

# Compare doc2 with doc3
similarity_doc2_doc3 = doc2.similarity(doc3)
print(f"Similarity between doc2 and doc3: {similarity_doc2_doc3}")
```

```python
doc1_100 = doc1[:100]
doc2_100 = doc2[:100]
doc3_100 = doc3[:100]

# Compare the first 100 tokens of doc1 with doc2
similarity_doc1_doc2_100 = doc1_100.similarity(doc2_100)
print(f"Similarity between the first 100 tokens of doc1 and doc2: {similarity_doc1_doc2_100}")

# Compare the first 100 tokens of doc1 with doc3
similarity_doc1_doc3_100 = doc1_100.similarity(doc3_100)
print(f"Similarity between the first 100 tokens of doc1 and doc3: {similarity_doc1_doc3_100}")

# Compare the first 100 tokens of doc2 with doc3
similarity_doc2_doc3_100 = doc2_100.similarity(doc3_100)
print(f"Similarity between the first 100 tokens of doc2 and doc3: {similarity_doc2_doc3_100}")
```

```python
# Import the Doc and Span classes
from spacy.tokens import Doc, Span

words = ["Margarita", "Stoyanova", "is", "a", "student", "at", "KdG", "."]
spaces = [True, True, True, True, True, True, False, False]

# Create a doc from the words and spaces
doc = Doc(nlp.vocab, words=words, spaces=spaces)
print(doc.text)


span = Span(doc, 0, 2, label="PERSON")
print(span.text, span.label_)

# Add the span to the doc's entities
doc.ents = [span]

# Print entities' text and labels
print([(ent.text, ent.label_) for ent in doc.ents])
```
