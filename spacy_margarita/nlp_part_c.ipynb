```python
import spacy
from spacy.language import Language

# Create a blank spaCy model
nlp = spacy.blank("en")

# Step 1: Print the pipeline components (it should be empty initially for a blank model)
print("Pipeline components before adding custom component:", nlp.pipe_names)

# Step 2: Define and register the custom component using the @Language.component decorator
@Language.component("longest_token_component")
def longest_token_component(doc):
    # Find the longest token
    longest_token = max(doc, key=len)
    print(f"The longest token in the document is: '{longest_token}'")
    return doc  # Must return the doc object

# Add the custom component to the pipeline using its registered name
nlp.add_pipe("longest_token_component", name="longest_token_component")

# Print the pipeline components again to see the custom component added
print("Pipeline components after adding custom component:", nlp.pipe_names)

# Test with a sample text
doc = nlp("spaCy is a great library for Natural Language Processing!")
```
