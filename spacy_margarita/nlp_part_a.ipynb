```python
import spacy
from spacy.matcher import Matcher
# Load the small English pipeline
nlp = spacy.load("en_core_web_sm")

```

```python

# Load docs with proper encoding handling
with open('moby_dick.txt', 'r', encoding='utf-8', errors='ignore') as file:
    moby_dick = file.read()

with open('ai_forecast1.txt', 'r', encoding='utf-8', errors='ignore') as file:
    ai_f1 = file.read()
    
with open('ai_forecast2.txt', 'r', encoding='utf-8', errors='ignore') as file:
    ai_f2 = file.read()

```

```python
# Process a text
doc_moby = nlp(moby_dick)
doc_1 = nlp(ai_f1)
doc_2 = nlp(ai_f2)
```

```python
def analyze_text(doc):
    print("\n---- Tokenization, POS, and Grammatical Structure ----")
    for token in doc:
       
        print(f"Token: {token.text}, POS: {spacy.explain(token.pos_)}, Dependency: {spacy.explain(token.dep_)}, Head: {token.head.text}")

    print("\n---- Sentence Segmentation ----")
    for sent in doc.sents:
        print(f"Sentence: {sent.text}\n-")

    print("\n---- Named Entities ----")
    for ent in doc.ents:
        print(f"Entity: {ent.text}, Label: {ent.label_}")
        
```

```python
print("### Analysis of Moby Dick ###")
analyze_text(doc_moby)
```

```python
print("### Analysis of Document 1 ###")
analyze_text(doc_1)
```

```python
print("### Analysis of Document 2 ###")
analyze_text(doc_2)
```

```python
# Initialize the Matcher
matcher_pattern_ai = Matcher(nlp.vocab)
# Define a pattern for "AI" followed by a verb

pattern_ai = [{"LOWER": "artificial"}, {"LOWER": "intelligence"}]

# Add pattern to the matcher
matcher_pattern_ai.add("ARTIFICIAL_INTELLIGENCE", [pattern_ai])

# Find matches
matches_doc1 = matcher_pattern_ai(doc_1)
matches_doc2 = matcher_pattern_ai(doc_2)

print("Matches for 'Artificial Intelligence' in doc 1:")
for match_id, start, end in matches_doc1:
    matched_span = doc_1[start:end]
    print(matched_span.text)

print("Matches for 'Artificial Intelligence' in doc 2:")
for match_id, start, end in matches_doc2:
    matched_span = doc_2[start:end]
    print(matched_span.text)
```

```python
pattern_ai_verb = [{"LOWER": "ai"}, {"POS": "VERB"}]
matcher_ai_verb = Matcher(nlp.vocab)
# Add the pattern to the matcher
matcher_ai_verb.add("AI_VERB", [pattern_ai_verb])

# Find matches
matches_doc1 = matcher_ai_verb(doc_1)
matches_doc2 = matcher_ai_verb(doc_2)

print("\nMatches for 'AI' followed by a verb in doc 1:")
for match_id, start, end in matches_doc1:
    matched_span = doc_1[start:end]
    print(matched_span.text)

print("\nMatches for 'AI' followed by a verb in doc 2:")
for match_id, start, end in matches_doc2:
    matched_span = doc_2[start:end]
    print(matched_span.text)
```

```python
pattern_percent = [{"LIKE_NUM": True}, {"TEXT": "%"}]
matcher_percent = Matcher(nlp.vocab)
# Add the pattern to the matcher
matcher_percent.add("PERCENT", [pattern_percent])


# Find matches
matches_doc1 = matcher_percent(doc_1)
matches_doc2 = matcher_percent(doc_2)

print("\nMatches for numbers followed by '%' in doc 1:")
for match_id, start, end in matches_doc1:
    matched_span = doc_1[start:end]
    print(matched_span.text)
    
print("\nMatches for numbers followed by '%' in doc 2:")
for match_id, start, end in matches_doc2:
    matched_span = doc_2[start:end]
    print(matched_span.text)
 
```

```python
# companies with matcher (separates multi-word entities)
org_pattern = [{"ENT_TYPE": "ORG"}]
matcher_org = Matcher(nlp.vocab)
matcher_org.add("ORG_PATTERN", [org_pattern])




matches_doc1 = matcher_org(doc_1)
matches_doc2 = matcher_org(doc_2)

print("\nMatches for organizations in doc 1:")
for match_id, start, end in matches_doc1:
    matched_span = doc_1[start:end]
    print(matched_span.text)
    
print("\nMatches for organizations in doc 2:")
for match_id, start, end in matches_doc2:
    matched_span = doc_2[start:end]
    print(matched_span.text)
```

```python
# companies with NER
print("\nOrganization names found in doc 1:")
for ent in doc_1.ents:
    if ent.label_ == "ORG":
        print(ent.text)

# Extract and print organization names using NER for doc_2
print("\nOrganization names found in doc 2:")
for ent in doc_2.ents:
    if ent.label_ == "ORG":
        print(ent.text)
```
